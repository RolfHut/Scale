```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
# Read the CSV
df = pd.read_csv(
    "AWH Accra Account Rolf - Data.csv",
    parse_dates=[0],
    date_format="%d-%m-%Y %H:%M:%S"
)

# Rename the first column (optional)
df.rename(columns={df.columns[0]: "Time"}, inplace=True)

# Make time the index
df.set_index("Time", inplace=True)

print(df)
```

```python
# Plot all columns versus time

def plotAllColumns(dataframe):
    for column in dataframe.columns:
        plt.figure(figsize=(10, 4))
        plt.plot(dataframe.index, dataframe[column], linewidth=1)
        plt.title(column)
        plt.xlabel("Time")
        plt.ylabel(column)
        plt.grid(True)
        plt.tight_layout()
    
        plt.savefig(f"{column}.png", dpi=150)
        plt.show()
        plt.close()

plotAllColumns(df)
```

```python
# Select the last 24 hours
df_last12h = df.loc[df.index >= (df.index.max() - pd.Timedelta(hours=12))]
plotAllColumns(df_last12h)
```
